```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'Name'   : ['Ravi','Priya','Arjun','Sneha','Kiran','Meera','Raj','Divya','Suresh','Ananya'],
    'Marks'  : [85, 90, 78, 92, 88, 76, 95, 60, 55, 83],
    'Age'    : [21, 22, 21, 23, 22, 21, 23, 22, 21, 22],
    'City'   : ['Hyderabad','Vizag','Kakinada','Chennai','Hyderabad',
                'Vizag','Kakinada','Chennai','Hyderabad','Vizag'],
    'Gender' : ['M','F','M','F','M','F','M','F','M','F'],
    'Fees'   : [15000,18000,12000,20000,15000,11000,22000,9000,13000,17000]
}
df = pd.DataFrame(data)
print(df)
# Monthly sales data for line charts
months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
sales  = [12000,15000,11000,18000,22000,19000,25000,23000,27000,21000,29000,35000]
```

```python
# Draw a bar chart of student names vs marks
plt.figure(figsize=(10,5))
plt.bar(df['Name'],df['Marks'])
plt.title("student marks")
plt.xlabel("Names")
plt.ylabel("Marks")
plt.savefig("student_marks.png")
plt.show()
print()
# Draw a line chart showing monthly sales trend
plt.plot(months,sales)
plt.title("Monthly Sales Trend")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
print()
# Draw a horizontal bar chart of names vs fees paid
plt.barh(df['Name'],df['Fees'])
plt.title("Fees Paid")
plt.xlabel("Fees")
plt.ylabel("Names")
plt.show()
```

```python
# Draw a histogram of marks distribution
plt.hist(df['Marks'])
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
print()
# Draw a scatter plot of Age vs Marks
plt.scatter(df['Age'],df['Marks'])
plt.title("Age vs Marks")
plt.xlabel("Age")
plt.ylabel("Marks")
plt.show()
# Draw a pie chart of how many students are from each city
city_counts=df['City'].value_counts()
plt.pie(city_counts,labels=city_counts.index)
```

```python
# Count male vs female students and show as a bar chart using Seaborn
sns.countplot(x='Gender',data=df)
plt.show()
print()
# Draw a Seaborn bar chart of average marks per city
sns.barplot(x='City',y='Marks',data=df)
plt.show()
# Draw a box plot of marks — check for outliers
sns.boxplot(y='Marks',data=df)
plt.show()
```

```python
# Draw a scatter plot with Male/Female in different colors
sns.scatterplot(x='Age',y='Marks',hue='Gender',data=df)
plt.show()
print()
# Draw a Seaborn histogram of marks with KDE curve
sns.histplot(df['Marks'],kde=True)
plt.show()
```
