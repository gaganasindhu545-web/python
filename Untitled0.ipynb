```python
from numpy import nan
# pandas
# pandas libray is used to store the data.In this library we have many built in funtions these are used to store,analyze,explore,clean,and load the data
import pandas as pd
# creating a dictionary
data={
    'name':['sindhu','hari','deepthi','kalyani'],
    'age':[19,20,15,20],
    'marks':[80,99,89,nan]
}
df=pd.DataFrame(data)
print(df)
# built in functions in pandas
print(df.head(2))   # displaying first n rows
print(df.tail(2))   # displaying last n rows
print(df.shape)     # no of rows and columns it shows how big your data is
print(df.info())    # gives count of null values,columns with datatypes
print(df.describe()) # displaying statistics
print(df.isnull)
print(df.dropna())   # remove row with missing values
print(df.dtypes)     # display datatype with each column
print(df['age']>17)  #filtering by condition
print(df['name'])    # displaying specific column
print(df.rename(columns={'name':'names'}))
print(df.rename(columns={'age':'year'}))
print(df.fillna(90))  # replace the value
```

```python
# Find average marks of all students using df["Marks"].mean()
print(df["marks"].mean())
print(df['age'].max())
print(df['age'].min())
print(df['marks'].std())
# Filter students who scored more than 85 marks
print(df[df['marks']>85]['name'])
print(df.sort_values(by='marks',ascending=False))
```

```python
import pandas as pd

data = {
    'Name'     : ['Ravi','Priya','Arjun','Sneha','Kiran','Meera','Raj','Divya','Suresh','Ananya'],
    'Age'      : [21, 22, 21, 23, 22, 21, 23, 22, 21, 22],
    'Marks'    : [85, 90, 78, 92, 88, 76, 95, 60, 55, 83],
    'City'     : ['Hyderabad','Vizag','Kakinada','Chennai','Hyderabad',
                  'Vizag','Kakinada','Chennai','Hyderabad','Vizag'],
    'Gender'   : ['M','F','M','F','M','F','M','F','M','F'],
    'Fees_Paid': [True,True,False,True,True,False,True,False,True,True]
}
df = pd.DataFrame(data)
# finding average of all students
print(df['Marks'].mean())
# Filter students who scored more than 85 marks
print(df[df['Marks']>85]['Name'])
# Sort students by marks in descending order
print(df.sort_values(by='Marks',ascending=False))
# Find which city has the highest average marks
print(df.groupby('City')['Marks'].mean())
# Add a new column "Grade" — A if marks >= 90, else B
df['Grade']=df['Marks'].apply(lambda x:'A' if x>=90 else 'B')
print(df)
# Count how many students are from each city
print(df.groupby('City')['Name'].value_counts())
# Find the student with the highest marks
print(df.loc[df['Marks'].idxmax(),'Name'])
# Show only Name and Marks columns
print(df[['Name','Marks']])
# Find total number of students in the dataset
print(df['Name'].count())
# Check if there are any missing values in the data
print(df.isnull().sum())
# Find average marks separately for Male and Female students
print(df.groupby('Gender')['Marks'].mean())
#	Filter students who have NOT paid fees
print(df[df['Fees_Paid']==False])
# 13	Add a column "Result" — Pass if marks >= 60, else Fail
df['Result']=df['Marks'].apply(lambda x:'Pass' if x>=60 else 'Fail')
print(df)
# Find the top 3 students by marks
print(df.nlargest(3,'Marks'))
# Find how many students passed and failed
print(df.groupby('Result')['Name'].count())
# Find minimum and maximum marks in each city
print(df.groupby('City')['Marks'].agg(['min','max']))
# Rename column "Marks" to "Score"
print(df.rename(columns={'Marks':'Score'}))
# Select students aged 22 from Vizag
print(df[(df['Age']==22) & (df['City']=='Vizag')])
# Find the average age of students per city
print(df.groupby('City')['Age'].mean())
# Drop the "Fees_Paid" column from the DataFrame
print(df.drop(columns=['Fees_Paid'],inplace=True))
print(df)
```
